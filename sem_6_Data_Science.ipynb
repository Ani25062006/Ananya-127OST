```python
# 3.2 Describing the Data
import pandas as pd

url = "http://tinyurl.com/titanic-csv"
df = pd.read_csv(url)

# Show first two rows
print(df.head(2))

# Show dimensions
print("Dimensions: {}".format(df.shape))

# Show statistics
print(df.describe())
```

```python
# 3.3 Navigating DataFrames

# Select the first row
print(df.iloc[0])

# Select three rows (rows 1, 2, 3)
print(df.iloc[1:4])

# All rows up to and including the fourth row
print(df.iloc[:4])

# Set index as 'Name'
df = df.set_index(df['Name'])
print(df.loc['Allen, Miss Elisabeth Walton'])
```

```python
# 3.4 Selecting Rows Based on Conditionals

# Select top two rows where column 'Sex' is 'female'
print(df[df['Sex'] == 'female'].head(2))

# Multiple conditions: 'female' and 'Age' >= 65
print(df[(df['Sex'] == 'female') & (df['Age'] >= 65)])
```

```python
# 3.5 Replacing Values

# Replace any instance of 'female' with 'Woman'
print(df['Sex'].replace('female', 'Woman').head(2))

# Replace multiple values
print(df['Sex'].replace({'female': 'Woman', 'male': 'Man'}).head(5))

# Replace numeric values (example: 1 -> "One")
print(df.replace(1, "One").head(2))
```

```python
# 3.6 Renaming Columns

# Rename a single column
print(df.rename(columns={'PClass': 'Passenger Class'}).head(2))

# Rename multiple columns
print(df.rename(columns={'PClass': 'Passenger Class', 'Sex': 'Gender'}).head(2))
```

```python
# 3.7 Finding Min, Max, Sum, Average, and Count

print('Maximum:', df['Age'].max())
print('Minimum:', df['Age'].min())
print('Mean:', df['Age'].mean())
print('Sum:', df['Age'].sum())
print('Count:', df['Age'].count())

# Apply to whole DataFrame
print("Variance:\n", df.var(numeric_only=True))
print("Standard Deviation:\n", df.std(numeric_only=True))
print("Kurtosis:\n", df.kurt(numeric_only=True))
print("Skewness:\n", df.skew(numeric_only=True))
```

```python
# 3.8 Finding Unique Values

# Unique values
print(df['Sex'].unique())

# Value counts
print(df['Sex'].value_counts())
```

```python
# 3.9 Handling Missing Values

# Select missing values, show 2 rows
print(df[df['Age'].isnull()].head(2))
```

```python
# 3.10 Deleting a Column

# Drop column 'Age'
print(df.drop('Age', axis=1).head(2))
```

```python
# 3.11 Deleting a Row

# Exclude rows where 'Sex' is 'male'
print(df[df['Sex'] != 'male'].head(2))

# Delete a row by unique value in 'Name'
print(df[df['Name'] != 'Allison, Miss Helen Loraine'].head(2))

# Delete a row by index
print(df[df.index != 'Allen, Miss Elisabeth Walton'].head(2))
```

```python
# 3.13 Grouping Rows by Values

# Group by Sex
print(df.groupby('Sex').mean(numeric_only=True))

# Count how many survived
print(df.groupby('Survived')['Name'].count())

# Mean age by sex and survival
print(df.groupby(['Sex', 'Survived'])['Age'].mean())
```

```python
# 3.15 Looping Over a Column

# For loop
for name in df['Name'][0:2]:
    print(name.upper())

# List comprehension
print([name.upper() for name in df['Name'][0:2]])
```

```python
# 3.16 Applying a Function Over All Elements in a Column

def uppercase(x):
    return x.upper()

print(df['Name'].apply(uppercase)[0:2])
```

```python
# 3.17 Applying a Function to Groups

print(df.groupby('Sex').apply(lambda x: x.count()))
```

```python
# 3.18 Concatenating DataFrames

df1 = df.head(3)
df2 = df.tail(3)

# Concatenate vertically (stacking)
result = pd.concat([df1, df2])
print(result)
```

```python
# ============================================================
#               DATA PREPROCESSING PRACTICAL
# ============================================================

import numpy as np
import pandas as pd
from sklearn.preprocessing import MinMaxScaler, StandardScaler, RobustScaler, Normalizer
from sklearn.cluster import KMeans
from sklearn.datasets import make_blobs

# ============================================================
# 4.1 Rescaling a Feature (MinMax Scaler)
# ============================================================
print("4.1 — Rescaling a Feature (MinMaxScaler)\n")

feature = np.array([
    [-500.5],
    [-100.1],
    [0],
    [100.1],
    [900.9]
])

minmax_scaler = MinMaxScaler(feature_range=(0, 1))
scaled_feature = minmax_scaler.fit_transform(feature)

print("Original Feature:\n", feature)
print("\nScaled Feature:\n", scaled_feature)
print("\n------------------------------------------\n")

# ============================================================
# 4.2 Standardizing a Feature (StandardScaler)
# ============================================================
print("4.2 — Standardizing Feature (StandardScaler)\n")

feature2 = np.array([
    [-1000.1],
    [-200.2],
    [500.5],
    [600.6],
    [9000.9]
])

scaler = StandardScaler()
standardized = scaler.fit_transform(feature2)

print("Standardized Feature:\n", standardized)
print("Mean:", standardized.mean())
print("Standard Deviation:", standardized.std())
print("\n------------------------------------------\n")

# ============================================================
# RobustScaler
# ============================================================
print("4.2 — RobustScaler\n")

robust_scaler = RobustScaler()
robust_scaled = robust_scaler.fit_transform(feature2)

print("Robust Scaled Feature:\n", robust_scaled)
print("\n------------------------------------------\n")

# ============================================================
# 4.3 Normalizing Observations
# ============================================================
print("4.3 — Normalizing Observations\n")

features = np.array([
    [0.5, 0.5],
    [1.1, 3.4],
    [1.5, 20.2],
    [1.63, 34.4],
    [10.9, 3.3]
])

normalizer = Normalizer(norm="l2")
normalized = normalizer.fit_transform(features)

print("L2 Normalized:\n", normalized)

features_l1 = Normalizer(norm="l1").fit_transform(features)
print("\nL1 Normalized First Row Sum:", features_l1[0].sum())

print("\n------------------------------------------\n")

# ============================================================
# 4.9 Clustering (KMeans)
# ============================================================
print("4.9 — Clustering with KMeans\n")

features_cluster, _ = make_blobs(
    n_samples=50,
    n_features=2,
    centers=3,
    random_state=1
)

df = pd.DataFrame(features_cluster, columns=["feature_1", "feature_2"])

clusterer = KMeans(n_clusters=3, random_state=0)
df["group"] = clusterer.fit_predict(features_cluster)

print(df.head())
print("\n------------------------------------------\n")

# ============================================================
# 4.10 Deleting Missing Values
# ============================================================
print("4.10 — Delete missing values\n")

features_missing = np.array([
    [1.1, 11.1],
    [2.2, 22.2],
    [3.3, 33.3],
    [np.nan, 55]
])

# Remove rows with missing values
cleaned = features_missing[~np.isnan(features_missing).any(axis=1)]
print("After Removing Missing:\n", cleaned)

df_missing = pd.DataFrame(features_missing, columns=["feature_1", "feature_2"])
print("\nUsing pandas dropna():\n", df_missing.dropna())

print("\n------------------------------------------\n")

# ============================================================
# 4.11 Imputing Missing Values
# ============================================================
print("4.11 — Imputing Missing Values\n")

from sklearn.impute import SimpleImputer

features_imp, _ = make_blobs(
    n_samples=100,
    n_features=2,
    random_state=1
)

scaler = StandardScaler()
standardized_features = scaler.fit_transform(features_imp)

true_value = standardized_features[0, 0]
standardized_features[0, 0] = np.nan

mean_imputer = SimpleImputer(strategy="mean")
features_mean_imputed = mean_imputer.fit_transform(standardized_features)

print("True Value:", true_value)
print("Imputed Value:", features_mean_imputed[0, 0])
```

```python
import tkinter as tk
from tkinter import scrolledtext
import numpy as np
import pandas as pd
from sklearn.preprocessing import MinMaxScaler, StandardScaler, RobustScaler, Normalizer
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans
from sklearn.impute import SimpleImputer

# -------------------- GUI FUNCTION DEFINITIONS --------------------

def rescale_feature():
    feature = np.array([[-500.5], [-100.1], [0], [100.1], [900.9]])
    scaler = MinMaxScaler(feature_range=(0, 1))
    scaled = scaler.fit_transform(feature)
    output_box.delete(1.0, tk.END)
    output_box.insert(tk.END, "Rescaling (MinMaxScaler):\n")
    output_box.insert(tk.END, scaled)

def standardize_feature():
    feature = np.array([[-1000.1], [-200.2], [500.5], [600.6], [9000.9]])
    scaler = StandardScaler()
    standardized = scaler.fit_transform(feature)
    output_box.delete(1.0, tk.END)
    output_box.insert(tk.END, "Standardizing (StandardScaler):\n")
    output_box.insert(tk.END, f"{standardized}\n\nMean: {standardized.mean()}\nStd Dev: {standardized.std()}")

def robust_scaling():
    feature = np.array([[-1000.1], [-200.2], [500.5], [600.6], [9000.9]])
    scaler = RobustScaler()
    robust = scaler.fit_transform(feature)
    output_box.delete(1.0, tk.END)
    output_box.insert(tk.END, "Robust Scaling:\n")
    output_box.insert(tk.END, robust)

def normalization():
    features = np.array([
        [0.5, 0.5],
        [1.1, 3.4],
        [1.5, 20.2],
        [1.63, 34.4],
        [10.9, 3.3]
    ])

    normalizer = Normalizer(norm="l2")
    l2 = normalizer.transform(features)

    output_box.delete(1.0, tk.END)
    output_box.insert(tk.END, "Normalization (L2):\n")
    output_box.insert(tk.END, f"{l2}")

def kmeans_clustering():
    features, _ = make_blobs(n_samples=50, n_features=2, centers=3, random_state=1)
    clusterer = KMeans(n_clusters=3, random_state=0)
    groups = clusterer.fit_predict(features)

    df = pd.DataFrame(features, columns=["feature_1", "feature_2"])
    df['group'] = groups

    output_box.delete(1.0, tk.END)
    output_box.insert(tk.END, "KMeans Clustering (first 10 rows):\n")
    output_box.insert(tk.END, df.head(10))

def delete_missing_values():
    features = np.array([
        [1.1, 11.1],
        [2.2, 22.2],
        [3.3, 33.3],
        [np.nan, 55]
    ])
    cleaned = features[~np.isnan(features).any(axis=1)]
    output_box.delete(1.0, tk.END)
    output_box.insert(tk.END, "After Deleting Missing Values:\n")
    output_box.insert(tk.END, cleaned)

def impute_missing_values():
    features, _ = make_blobs(n_samples=100, n_features=2, random_state=1)
    scaler = StandardScaler()
    std = scaler.fit_transform(features)

    true_value = std[0, 0]
    std[0, 0] = np.nan

    imputer = SimpleImputer(strategy="mean")
    imputed = imputer.fit_transform(std)

    output_box.delete(1.0, tk.END)
    output_box.insert(tk.END, "Imputation of Missing Value:\n")
    output_box.insert(tk.END, f"True Value: {true_value}\nImputed Value: {imputed[0, 0]}")

# -------------------- GUI DESIGN --------------------

root = tk.Tk()
root.title("Machine Learning Preprocessing GUI")
root.geometry("700x600")
root.config(bg="#ECECEC")

title = tk.Label(root, text="ML Preprocessing Toolkit", font=("Arial", 18, "bold"), bg="#ECECEC")
title.pack(pady=10)

frame = tk.Frame(root, bg="#ECECEC")
frame.pack()

buttons = [
    ("Rescale Feature", rescale_feature),
    ("Standardize Feature", standardize_feature),
    ("Robust Scaling", robust_scaling),
    ("Normalize Observations", normalization),
    ("KMeans Clustering", kmeans_clustering),
    ("Delete Missing Values", delete_missing_values),
    ("Impute Missing Values", impute_missing_values)
]

for text, cmd in buttons:
    tk.Button(frame, text=text, width=25, height=2, command=cmd).pack(pady=5)

output_box = scrolledtext.ScrolledText(root, width=80, height=15, font=("Courier", 10))
output_box.pack(pady=10)

root.mainloop()
```
